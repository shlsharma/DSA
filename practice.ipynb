```python
def binary_search(nums, n):
    left, right = 0, len(nums)-1
    while left < right:
        mid = (left+right)//2
        if n < nums[mid]:
            right = mid
        elif n > nums[mid]:
            left = mid
        else:
            print('Found!!')
            return True
    print('Not Found !!')
    return False

binary_search([1, 2, 3, 4, 5, 6, 7, 8, 9], 3)
#              0, 1, 2, 3, 4, 5, 6, 7, 8
```

```python
def max_volume(heights):
    max_vol = 0
    left, right = 0, len(heights)-1
    res = (left, right)

    while left < right:
        vol = min(heights[left], heights[right]) * (right-left)
        if vol > max_vol:
            max_vol = vol
            res = (left, right)
        if heights[left] <= heights[right]:
            left += 1
        elif heights[left] > heights[right]:
            right -= 1
    return max_vol, res
            

max_volume([1,8,100,2,80,4,8,3,7])
```

```python
def trap_water(heights):
    left, right = 0, len(heights)-1
    left_max = 0
    right_max = 0
    water = 0

    while left < right:
        if heights[left] <= heights[right]:
            if left_max < heights[left]:
                left_max = heights[left]
            else:
                water += left_max-heights[left]
            left += 1
        else:
            if right_max < heights[right]:
                right_max = heights[right]
            else:
                water += right_max - heights[right]
            right -= 1
    return water

trap_water([0,1,0,2,1,0,1,3,2,1,2,1])
```

```python
def merge_intervals(intervals):
    if not intervals:
        return -1
    
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]

    for current in intervals[1:]:
        last = merged[-1]
        if current[0] <= last[1]:
            merged[-1][1] = max(last[1], current[1])
        else:
            merged.append(current)
    
    return merged

arr = [[0,0],[1,2],[5,5],[2,4],[3,3],[5,6],[5,6],[4,6],[0,0],[1,2],[0,2],[4,5]]
merge_intervals(arr)
```

```python
#18 Valid Palindrome (LC 125)
def valid_palindrome(string):
    ns = ''.join(s.lower() for s in string if s.isalnum())
    left, right = 0, len(ns)-1
    while left<right:
        if ns[left] != ns[right]:
            return False
        else:
            left+=1
            right-=1
    return True

valid_palindrome('A man, a plan, a canal: Panama')
```

```python
#19 longest common preffix LC 14
def longest_common_prefix(strs):
    if not strs:
        return ""
    
    prefix = strs[0]

    for s in strs[1:]:
        while not s.startswith(prefix):
            prefix = prefix[:-1]
            if prefix == '':
                return ''
    return prefix

longest_common_prefix(['flower', 'flow', 'floer'])
```

```python
#24 Leetcode 443 - String Compression

def string_compression(chars):
    read = 0
    write = 0 
    while read < len(chars):
        char = chars[read]
        count = 0 

        while read<len(chars) and char == chars[read]:
            read += 1
            count += 1
        
        chars[write] = char
        write+=1

        if count>1:
            for s in str(count):
                chars[write] = s
                write+=1

    return write

string_compression(["a","b","b","b","b","b","b","b","b","b","b","b","b"]) 
# [a, b, 1, 2]
```

```python
#25 Isomorphic String

def is_isomorphic(s, t):
    mapping_s_t = {}
    mapping_t_s = {}
    for i, j in zip(s, t):
        if (i not in mapping_s_t) and (j not in mapping_t_s):
            mapping_s_t[i] = j
            mapping_t_s[j] = i
        elif mapping_s_t[i] != j or mapping_t_s[j] != i:
            return False
    return True

is_isomorphic("abb", "xyx")
```

```python
#26 LC 20. Valid Parentheses
def valid_parentheses(s: str) -> bool:
    stack = []
    mapping = {')': '(', '}': '{', ']':'['}

    for p in s:
        if p not in mapping:
            stack.append(p)
        else:
            if not stack or stack[-1] != mapping[p]:
                return False
            stack.pop()
    return True

valid_parentheses('{[()]}[]')
```

```python
#28 Leetcode 1 - Two Sum
# arr = [2, 7, 8, 9] target=9
def twoSum(arr, target):
    seen = {}
    for idx, num in enumerate(arr):
        if num in seen:
            return (seen[num], idx)
        seen[target-num] = idx
    return -1

twoSum([2, 7, 11, 15], 9)
```

```python
#31 LeetCode 349: Find the array intersection
def find_array_intersection(num1, num2):
    num1_set = set(num1)
    num2_set = set(num2)
    result = []
    for num in num1_set:
        if num in num2_set:
            result.append(num)
    return result

find_array_intersection([4,9,5], [9,4,9,8,4])
```

```python
#32 LC 202: Happy Number
def is_happy_number(num):
    seen = {}
    while True:
        num = sum([int(i)**2 for i in str(num)])
        if num==1:
            return True
        elif num not in seen:
            seen[num] = seen.get(num, 0) + 1
        else:
            return False
        
is_happy_number(18)
```

```python
#33 LC 387: find first unique character in the string
from collections import Counter
def first_unique_char(s):
    count = Counter(s)
    for idx, char in enumerate(s):
        if count[char] == 1:
            return idx
    return -1

first_unique_char('aabbcdeb')
```

```python
#34 LC 347: Top K Frequent Elements
from collections import Counter

def k_frequent_elements(nums, k):
    count = Counter(nums)
    sorted_count = sorted(count.items(), key=lambda x: x[1], reverse=True)
    return [i[0] for i in sorted_count][:k]

k_frequent_elements([1, 1, 1, 2, 2, 2, 2, 2, 3], 2)
```
