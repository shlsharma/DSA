```python
def binary_search(nums, n):
    left, right = 0, len(nums)-1
    while left < right:
        mid = (left+right)//2
        if n < nums[mid]:
            right = mid
        elif n > nums[mid]:
            left = mid
        else:
            print('Found!!')
            return True
    print('Not Found !!')
    return False

binary_search([1, 2, 3, 4, 5, 6, 7, 8, 9], 3)
#              0, 1, 2, 3, 4, 5, 6, 7, 8
```

```python
def find_min_max(nums):
    curr_min = float('inf')
    curr_max = float('-inf')

    for num in nums:
        if num < curr_min:
            curr_min = num
        if num > curr_max:
            curr_max = num
    return curr_min, curr_max

find_min_max([3, 4, 5, 2, 4, 3, 45])
```

```python
def preffix_sum(nums):

    for i in range(1, len(nums)):
        nums[i] += nums[i-1]
    
    return nums

preffix_sum([1, 2, 3, 4, 5, 6, 7, 8, 9])
```

```python
def suffix_sum(nums):

    for i in range(len(nums)-2, -1, -1):
        nums[i] += nums[i+1]
    
    return nums

suffix_sum([1,2,3,4,5,6,7,8,9])
```

```python
def pair_with_target_sum(nums, target):
    nums = sorted(nums)
    left, right = 0, len(nums)-1

    while left<right:
        curr_sum = nums[left]+nums[right]
        if curr_sum == target:
            return True, nums[left], nums[right]
        elif curr_sum <= target:
            left += 1
        elif curr_sum > target:
            right -= 1
    return False

pair_with_target_sum([1, 2, 3, 4, 5, 6], 5)
```

```python
def remove_duplicates(nums):
    count = 0

    for i in range(len(nums)):
        if nums[i] != nums[count]:
            count+=1
            nums[count] = nums[i]
    
    return nums[:count+1]

remove_duplicates([1,1,2,2,2,3,4,4,5,6])
```

```python
def reverse_array(nums):
    left, right = 0, len(nums)-1

    while left<right:
        nums[left], nums[right] = nums[right], nums[left]
        left += 1
        right -= 1
    
    return nums

reverse_array([1, 2, 3, 4, 5, 6])
```

```python
def check_palindrome(string):
    left, right = 0, len(string)-1

    while left<right:
        if string[left] == string[right]:
            left += 1
            right -= 1
        else:
            return False
    return True

check_palindrome('allma')
```

```python
def max_volume(heights):
    max_vol = 0
    left, right = 0, len(heights)-1
    res = (left, right)

    while left < right:
        vol = min(heights[left], heights[right]) * (right-left)
        if vol > max_vol:
            max_vol = vol
            res = (left, right)
        if heights[left] <= heights[right]:
            left += 1
        elif heights[left] > heights[right]:
            right -= 1
    return max_vol, res
            

max_volume([1,8,100,2,80,4,8,3,7])
```

```python
def move_zero(nums):
    count = 0

    for i in range(len(nums)):
        if nums[i] != 0:
            nums[count] = nums[i]
            count += 1
    nums[count:] = [0]*(len(nums)-count)
    return nums

move_zero([0, 1, 0, 3, 0, 13]) 
```

```python
def trap_water(heights):
    left, right = 0, len(heights)-1
    left_max = 0
    right_max = 0
    water = 0

    while left < right:
        if heights[left] <= heights[right]:
            if left_max < heights[left]:
                left_max = heights[left]
            else:
                water += left_max-heights[left]
            left += 1
        else:
            if right_max < heights[right]:
                right_max = heights[right]
            else:
                water += right_max - heights[right]
            right -= 1
    return water

trap_water([0,1,0,2,1,0,1,3,2,1,2,1])
```

```python
def merge_intervals(intervals):
    if not intervals:
        return -1
    
    intervals.sort(key=lambda x: x[0])
    merged = [intervals[0]]

    for current in intervals[1:]:
        last = merged[-1]
        if current[0] <= last[1]:
            merged[-1][1] = max(last[1], current[1])
        else:
            merged.append(current)
    
    return merged

arr = [[0,0],[1,2],[5,5],[2,4],[3,3],[5,6],[5,6],[4,6],[0,0],[1,2],[0,2],[4,5]]
merge_intervals(arr)
```

```python
def longest_substring(string):
    memo={}
    max_length = 0
    start, end = 0, 0

    while end < len(string):
        
        if string[end] in memo and memo[string[end]]>=start:
            start = memo[string[end]]+1
        
        memo[string[end]] = end
        max_length = max(max_length, end-start+1)
        end+=1
    return max_length

longest_substring('avniajjabcdefghijk')
```
