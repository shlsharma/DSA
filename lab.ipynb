```python
def binary_search(array, target):
    left, right = 0, len(array)-1

    while left <= right:
        mid = (left+right)//2

        if array[mid] == target:
            return mid
        elif array[mid] <= target:
            left = mid + 1
        else:
            right = mid - 1

    return -1
```

```python
#29 Leetcode - 560: Subarray sum equals to K
# nums = [1, 1, 1, 2] target=3
# ans = (1, 1, 1), (1, 2) = 2
'''
🔁 Goal:
We want to know:

Why does prefix_sum[i] - k in seen mean there exists a subarray that sums to k?

📘 Let's Define:
prefix_sum[i] = sum of elements from index 0 to i

prefix_sum[j] = sum of elements from index 0 to j

Then the sum of elements between j+1 and i is:

python
Copy
Edit
prefix_sum[i] - prefix_sum[j]
So if:

python
Copy
Edit
prefix_sum[i] - prefix_sum[j] = k
Then:

python
Copy
Edit
prefix_sum[j] = prefix_sum[i] - k
So:

If at position i, the running sum is prefix_sum[i], and we’ve seen a prefix sum equal to prefix_sum[i] - k at some earlier index j, then the subarray (j+1 to i) has sum k.
'''
def subarraySum(nums, k):
    seen = {0: 1}  # prefix_sum 0 occurs once
    cum_sum = 0
    count = 0

    for num in nums:
        print('seen: ', seen)
        cum_sum += num
        print(cum_sum, cum_sum-k)
        if cum_sum - k in seen:
            count += seen[cum_sum - k]
        seen[cum_sum] = seen.get(cum_sum, 0) + 1

    return count, seen


subarraySum([1, 2, 1, 2, 1], 3)
                
```

```python
#39 LC 438: Find anagrams in a string
from collections import Counter

def findAnagrams(s, p):
    if len(p)>len(s):
        return []
    
    count_p = Counter(p)
    window = Counter(s[:len(p)])
    result = []

    if count_p == window:
        result.append(0)

    for i in range(len(p), len(s)):
        window[s[i]] += 1
        window[s[i-len(p)]] -= 1

        if window[s[i-len(p)]] == 0:
            del window[s[i-len(p)]]
        
        if count_p == window:
            result.append(i-len(p)+1)
    return result

findAnagrams("abababa", "ab")
```

```python
#42 LC 447: Number of Boomerangs
from collections import defaultdict
def number_of_boomerangs(points):
    def calculate_dist(i, j):
        return ((i[0]-j[0])**2 + (i[1]-j[1])**2)**2
    
    dist_map = defaultdict(lambda: defaultdict(int))
    for point_i in points:
        for point_j in points:
            dist_map[tuple(point_i)][calculate_dist(point_i, point_j)]+=1
    return sum([j*(j-1) for i in dist_map.values() for j in i.values()])

number_of_boomerangs([[0,0],[1,0],[-1,0],[0,1],[0,-1]])
```

```python
#50 LC 46 Permutation
def permutation(nums):
    res = []

    def backtrace(path, used):
        if len(path) == len(nums):
            res.append(path[:])
            return
        for i in range(len(nums)):
            print('for loop: ', i, path, used)
            if used[i]:
                print('continue')
                continue
            path.append(nums[i])
            used[i] = True
            backtrace(path, used)
            print('before i: ',i, path, used)
            path.pop()
            used[i] = False
            print('after i: ',i, path, used)

    backtrace([], [0]*len(nums))
    return res

permutation([1, 2, 3])
```

```python
#55 selection sort 

def selection_sort(array):
    n = len(array)
    for i in range(n):
        min_index = i
        for j in range(i, n):
            if array[j] < array[min_index]:
                min_index = j
        if min_index!=i:
            array[i], array[min_index] = array[min_index], array[i]

    return array

selection_sort([2, 4, 5, 1, 3, 1, 1, 3, 4])
```

```python
#56 merge sort

def merge_sort(array):
    n = len(array)
    if n == 1 or n==0:
        return array

    mid = n//2
    left = array[:mid]
    right = array[mid:]

    left = merge_sort(left)
    right = merge_sort(right)

    l_len = len(left)
    r_len = len(right)

    merged_array = [0]*n

    l, r = 0, 0
    i = 0
    while l<l_len and r<r_len:
        if left[l]<right[r]:
            merged_array[i] = left[l]
            l+=1
        else:
            merged_array[i] = right[r]
            r+=1
        i+=1
    
    while l<l_len:
        merged_array[i] = left[l]
        l+=1
        i+=1
    
    while r<r_len:
        merged_array[i] = right[r]
        r+=1
        i+=1

    return merged_array

merge_sort([2, 4, 5, 1, 3, 1, 1, 3, 4])
```

```python
#55 Quick Sort

def quick_sort(array):
    n = len(array)

    if n == 0 or n == 1:
        return array
    
    pivot = array[-1]

    left = []
    right = []

    for i in range(n-1):
        if array[i]<=pivot:
            left.append(array[i])
        else:
            right.append(array[i])
    
    left = quick_sort(left)
    right = quick_sort(right)

    sorted_array = left + [pivot] + right

    return sorted_array

quick_sort([2, 4, 5, 1, 3, 1, 1, 3, 4])
```

```python
#56 Counting sort

def counting_sort(array):
    n = len(array)

    max_value = 0
    for i in range(n):
        if array[i] > max_value:
            max_value = array[i]
    
    counting_array = [0]*(max_value+1)
    for i in range(n):
        counting_array[array[i]] += 1
    c = 0
    for i in range(len(counting_array)):
        for j in range(counting_array[i]):
            array[c] = i
            c+=1

    return array

counting_sort([2, 4, 5, 1, 3, 1, 1, 3, 4])
```
